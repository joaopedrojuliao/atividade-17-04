**Exercício 1: Validador de Estoque e Segurança de Dados**
Enunciado: Imagine que você está lendo uma lista de produtos. Crie um programa que percorra uma lista de dicionários (contendo nome, quantidade e preço). Para cada item:

Se a quantidade for menor que 5 e o preço for maior que R$ 100,00, printe: "ALERTA: Reposição urgente para produto de alto valor: [Nome]".

Se a quantidade for 0, printe: "ERRO: Produto esgotado".

Use um operador booleano para ignorar produtos que tenham nomes inválidos (vazios ou apenas espaços).


```python
produtos = [
    {"nome": "Laptop", "quantidade": 3, "preço": 1200.00},
    {"nome": "Mouse", "quantidade": 10, "preço": 50.00},
    {"nome": "Teclado Gamer", "quantidade": 1, "preço": 150.00},
    {"nome": "Monitor 4K", "quantidade": 0, "preço": 400.00},
    {"nome": "Webcam HD", "quantidade": 7, "preço": 80.00},
    {"nome": "", "quantidade": 2, "preço": 200.00}, # Nome inválido
    {"nome": "   ", "quantidade": 4, "preço": 300.00}, # Nome inválido
    {"nome": "Fone de Ouvido", "quantidade": 0, "preço": 75.00}
]

for produto in produtos:
    # Operador booleano para ignorar produtos com nomes inválidos
    if not produto["nome"].strip():
        print(f"AVISO: Produto com nome inválido ignorado: {produto}")
        continue

    quantidade = produto["quantidade"]
    preco = produto["preço"]
    nome = produto["nome"]

    if quantidade < 5 and preco > 100.00:
        print(f"ALERTA: Reposição urgente para produto de alto valor: {nome}")
    elif quantidade == 0:
        print(f"ERRO: Produto esgotado: {nome}")
```

**Exercício 2: Simulador de Descontos Progressivos**
Enunciado: Uma loja aplica descontos baseados no faturamento total da compra e na região.
Crie um loop que processe uma lista de vendas. Aplique a seguinte lógica:

Se o valor for maior que 500 ou o cliente for da região "Norte", aplique 10% de desconto.

Se o valor for maior que 1000 e a região for "Sul", aplique 15% de desconto.

Se nenhuma condição for atendida, o desconto é 0.

O programa deve exibir o valor original, o desconto aplicado e o valor final.


```python
vendas = [
    {"valor": 600, "regiao": "Sudeste"},
    {"valor": 400, "regiao": "Norte"},
    {"valor": 1200, "regiao": "Sul"},
    {"valor": 800, "regiao": "Nordeste"},
    {"valor": 300, "regiao": "Centro-Oeste"},
    {"valor": 1500, "regiao": "Norte"},
    {"valor": 900, "regiao": "Sul"}
]

for venda in vendas:
    valor_original = venda["valor"]
    regiao = venda["regiao"]
    desconto_percentual = 0

    # Condição 1: Valor > 500 ou região "Norte" -> 10% de desconto
    if valor_original > 500 or regiao == "Norte":
        desconto_percentual = 0.10
    # Condição 2: Valor > 1000 e região "Sul" -> 15% de desconto
    # Esta condição deve ser verificada APÓS a primeira, pois pode ser mais específica
    # e precisa sobrepor a primeira se ambas forem verdadeiras para a mesma venda
    # A ordem das condições é importante para a lógica de 'elif'
    if valor_original > 1000 and regiao == "Sul":
        desconto_percentual = 0.15

    desconto_aplicado = valor_original * desconto_percentual
    valor_final = valor_original - desconto_aplicado

    print(f"Venda Original: R${valor_original:.2f}, Região: {regiao}")
    print(f"  Desconto Aplicado: R${desconto_aplicado:.2f} ({desconto_percentual * 100:.0f}%)")
    print(f"  Valor Final: R${valor_final:.2f}")
    print("----------------------------------------")
```

**Exercício 3: Sistema de Login e Bloqueio (Iteração com While)**
Enunciado: Simule um sistema de acesso aos dados da planilha.

O usuário tem 3 tentativas para digitar a senha correta ("python123").

Use um loop while e uma variável contadora.

Se ele acertar, mude uma variável booleana acesso_concedido para True e interrompa o loop com break.

Ao final, se acesso_concedido for False, exiba uma mensagem de "Conta Bloqueada".


```python
senha_correta = "python123"
tentativas_restantes = 3
acesso_concedido = False

while tentativas_restantes > 0:
    senha_digitada = input(f"Digite a senha ({tentativas_restantes} tentativas restantes): ")

    if senha_digitada == senha_correta:
        acesso_concedido = True
        print("Acesso concedido!")
        break
    else:
        tentativas_restantes -= 1
        if tentativas_restantes > 0:
            print("Senha incorreta. Tente novamente.")
        else:
            print("Você não tem mais tentativas.")

if not acesso_concedido:
    print("Conta Bloqueada")
```

**Exercício 4: Filtro de Vendedores de Elite**
Enunciado: Dada uma lista com os nomes dos vendedores e seus respectivos totais de vendas, utilize um comando iterativo para identificar:

Vendedores que venderam mais de R$ 5.000,00.

Mas atenção: Se o nome do vendedor for "Admin" ou "Teste", ele não deve ser contabilizado (use o comando continue).

Ao final, mostre a média de vendas apenas dos vendedores válidos que superaram a meta.



```python
vendedores = [
    {"nome": "Alice", "vendas": 6500.00},
    {"nome": "Bob", "vendas": 4800.00},
    {"nome": "Charlie", "vendas": 7200.00},
    {"nome": "Admin", "vendas": 9000.00},
    {"nome": "Teste", "vendas": 5500.00},
    {"nome": "Diana", "vendas": 5100.00},
    {"nome": "Eve", "vendas": 4900.00}
]

vendas_elite = []

print("Vendedores de Elite (não 'Admin' ou 'Teste' e vendas > R$5.000,00):")
for vendedor in vendedores:
    nome = vendedor["nome"]
    vendas = vendedor["vendas"]

    # Ignorar vendedores 'Admin' e 'Teste'
    if nome == "Admin" or nome == "Teste":
        continue

    # Identificar vendedores que venderam mais de R$ 5.000,00
    if vendas > 5000.00:
        vendas_elite.append(vendas)
        print(f"- {nome}: R${vendas:.2f}")

if vendas_elite:
    media_vendas_elite = sum(vendas_elite) / len(vendas_elite)
    print(f"\nMédia de vendas dos vendedores de elite válidos: R${media_vendas_elite:.2f}")
else:
    print("\nNenhum vendedor de elite válido encontrado.")
```
